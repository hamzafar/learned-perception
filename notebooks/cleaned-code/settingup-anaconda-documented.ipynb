Download the Miniconda installer and install it under `/opt/miniconda3`.

- `bash ... -b` runs the installer in batch mode without interactive prompts.
- `-p` sets the installation prefix.

The installation output is retained below as evidence of what happened in the recorded run.

```python
# Download the pinned Miniconda installer used for this environment.
!wget -q https://repo.anaconda.com/miniconda/Miniconda3-py38_23.11.0-2-Linux-x86_64.sh -O /tmp/miniconda.sh

# Install Miniconda non-interactively at a known path.
!bash /tmp/miniconda.sh -b -p /opt/miniconda3
```

## 3. Create the `bevfusion` Environment

Create a separate Conda environment with Python 3.8. Keeping the project dependencies isolated helps avoid conflicts with packages in Kaggle's default Python environment.

The command uses the Conda executable by its absolute path, so it does not depend on whether `conda` is available on the shell's `PATH`.

```python
# Create an isolated environment for BEVFusion and its legacy dependencies.
!/opt/miniconda3/bin/conda create -n bevfusion python=3.8 -y
```

## 4. Install PyTorch and Verify It

Install the specified PyTorch, torchvision, and torchaudio versions into the new environment, then print the PyTorch version using that environment's Python interpreter.

The `+cu113` suffix on PyTorch and torchvision indicates builds intended for CUDA 11.3. This version pin is part of the environment setup; later BEVFusion dependencies should be checked against the project's own compatibility requirements.

**Verification caveat:** The recorded output confirms that `torch` imports and reports `1.10.2+cu113`. It also shows `torchaudio` being installed as a `+rocm4.1` build, so this output alone does not establish that torchaudio is CUDA-compatible. Check that package if audio functionality or dependency consistency matters; do not assume it is needed by BEVFusion.

```python
# Install the pinned framework packages inside the dedicated environment.
!/opt/miniconda3/envs/bevfusion/bin/pip install torch==1.10.2+cu113 torchvision==0.11.3+cu113 torchaudio==0.10.2 -f https://download.pytorch.org/whl/torch_stable.html

# Confirm which PyTorch version imports in the bevfusion environment.
!/opt/miniconda3/envs/bevfusion/bin/python -c "import torch; print(torch.__version__)"
```
